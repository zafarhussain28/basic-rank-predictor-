```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.metrics import mean_absolute_error
import joblib
```

```python
df= pd.read_csv("/content/Realistic_JEE_Marks_Percentile_Rank_Dataset_v2.csv")
```

```python
df.loc[38]
```

```python
df = df[["raw_marks", "approx_percentile", "approx_rank"]].dropna()
```

```python
print(df.columns)
```

```python
X=df[['raw_marks']]
y_percentile=df['approx_percentile']
y_rank=df['approx_rank']
```

```python
X_train, X_test, y_train_p, y_test_p = train_test_split(
    X, y_percentile, test_size=0.2, random_state=42
)

# For rank
_, _, y_train_r, y_test_r = train_test_split(
    X, y_rank, test_size=0.2, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model1=LinearRegression()
model2=DecisionTreeRegressor()
model3= RandomForestRegressor(n_estimators=300, random_state=42)

models = {
    "Linear Regression": model1,
    "Decision Tree": model2,
    "Random Forest": model3
}

# Evaluation loop for Percentile
print("\n=== Percentile Prediction Results ===")

for name, model in models.items():
    model.fit(X_train, y_train_p)
    preds = model.predict(X_test)

    print(f"\nModel: {name}")
    print("MAE:", mean_absolute_error(y_test_p, preds))
    print("MSE:", mean_squared_error(y_test_p, preds))
    print("R2 :", r2_score(y_test_p, preds))
```

```python
print("\n=== Rank Prediction Results ===")

for name, model in models.items():
    model.fit(X_train, y_train_r)
    preds = model.predict(X_test)

    print(f"\nModel: {name}")
    print("MAE:", mean_absolute_error(y_test_r, preds))
    print("MSE:", mean_squared_error(y_test_r, preds))
    print("R2 :", r2_score(y_test_r, preds))
```

```python
print("\n=== Rank Prediction Results ===")

for name, model in models.items():
    model.fit(X_train, y_train_r)
    preds = model.predict(X_test)

    print(f"\nModel: {name}")
    print("MAE:", mean_absolute_error(y_test_r, preds))
    print("MSE:", mean_squared_error(y_test_r, preds))
    print("R2 :", r2_score(y_test_r, preds))
```

```python
import joblib

# Save Percentile Model
joblib.dump(models["Random Forest"].fit(X_train, y_train_p), "percentile_model.pkl")

# Save Rank Model
joblib.dump(models["Random Forest"].fit(X_train, y_train_r), "rank_model.pkl")

print("Models saved successfully!")
```
